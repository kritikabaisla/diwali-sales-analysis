```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Diwali Sales Data.csv",encoding="unicode_escape")
```

```python
df.drop(['Status','unnamed1'],axis=1,inplace=True)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.dropna(inplace=True)
```

```python
df.info()
```

```python
df['Amount']=df['Amount'].astype(int)
```

```python
df.rename(columns={'Marital_Status':'shaadi'})
```

```python
#up till here the cleaning steps are done
```

```python
df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
sns.countplot(x='Gender',data=df)
```

```python
ax=sns.countplot(x='Gender',data=df)
for bars in ax.containers:
  ax.bar_label(bars)
```

```python
sale_gen=df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Gender',y='Amount',data=sale_gen)
```

```python
#until here we understand that female purchases more compared to men and have purchased more items then men
```

```python
df.columns
```

```python
sns.countplot(x='Age Group',data=df,hue='Gender')
```

```python
age_gen=sns.countplot(x='Age Group',data=df,hue='Gender')
for bars in age_gen.containers:
  age_gen.bar_label(bars)
```

```python
#it shows that age group of 26-35 purchases the most with the female population
```

```python
df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
```

```python
agee=df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount',ascending=False)
sns.barplot(x='Age Group',y='Amount',data=agee)
```

```python
sns.countplot(x='State',data=df)
```

```python
df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
```

```python
sales_state=df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders',ascending=False).head(10)
sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(x='State',y='Orders',data=sales_state)
```

```python
#from here we understand that up state has the most orders.top3 are:up,maharashtra,karnatka
```
